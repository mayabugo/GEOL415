```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as colors
import xarray as xr
import cartopy.crs as ccrs
```

```python
ds = xr.open_dataset ('C:/Users/mayav/Desktop/GEOL415/small_GEOS_CF.nc')
```

```python
O3 = ds.O3 * 1e9 # Name dataset and multiply by a scalar that will convert data to ppb
```

```python
# Plot O3 levels on map
fig = plt.figure(figsize=(10,7))
ax = plt.axes(projection=ccrs.Robinson())
O3.plot(ax=ax, transform=ccrs.PlateCarree(), 
        cbar_kwargs={'shrink': 0.4})
ax.coastlines()
ax.gridlines()
```

```python
# Where is O3 qualitatively high? 
    # O3 looks high off the East coast of India, throughout sub-saharan Africa, and in the Eastern part of central America.
```

```python
O3.max()
```

```python
O3.where(O3==O3.max(), drop=True).squeeze() # "drop=true" removes 0 values & ".squeeze" makes value a scalar
```
